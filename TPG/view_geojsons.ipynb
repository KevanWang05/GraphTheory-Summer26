```python
#This is just to read in all of the geojsons
import json
import pandas as pd
import numpy as np
import geopandas as gpd
```

```python
nodes3 = gpd.read_file("C:/Users/kevan/OneDrive/Desktop/GraphTheory-Summer26/TPG/geojsons/3_nodes.geojson").set_index("graph_type_row")

# 1. Convert the DataFrame values to a pure NumPy array
matrix3 = nodes3.drop(columns=['geometry']).to_numpy()

# 2. Slice from the 2nd column (index 1) up to the 2nd-to-last column inclusive (index -1)
target_window = matrix3[:, 1:-1]

# 3. Check for 3s
# Returns True if there are no 3s in the window, and False if a 3 is found anywhere
result = not np.any(target_window == 3)

print(result)
```
